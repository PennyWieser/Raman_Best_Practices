```python
import numpy as np
import matplotlib.pyplot as plt

# Your spectra arrays: shape (n, 2) with [:,0]=x, [:,1]=y
datasets = {
    'Red'    : ('r', red),
    'Green'  : ('g', green),
    'Cyan'   : ('c', cyan),
    'Blue'   : ('b', blue),
    'Magenta': ('m', magenta),
}


fig, (ax1) = plt.subplots(1, 1, figsize = (3,4))

# Plotting loop
for label, (color, arr) in datasets.items():
    idx = np.argmin(np.abs(arr[:, 0] - 1380))
    y_offset = arr[idx, 1]
    x = arr[:, 0]
    y = arr[:, 1] - y_offset
    ax1.plot(x, y, '.-', color=color)

#ax1.legend(fontsize=8)
ax1.set_xlim([1380, 1395])
ax1.set_ylabel('Intensity (relative to bck @ 1380 cm$^{-1}$)')
ax1.set_xlabel('Raman shift (cm$^{-1}$)')
fig.tight_layout()
ax1.legend(fontsize=8)
#plt.ylim([-50, 650])
fig.savefig('Sectioning_diad2.png', dpi=300, transparent=True)
```
