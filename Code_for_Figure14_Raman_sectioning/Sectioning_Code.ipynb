```python
import numpy as np
import matplotlib.pyplot as plt

# Dictionary of label, color, and corresponding array
datasets = {
    '+2.3 µm': ('b', man_focus_112Z),
    '0 µm \n (oscilliscope)': ('magenta', man_focus_89Z),
     '-3.3 µm': ('c', man_focus_56Z),
       '-6.6 µm': ('r', man_focus_23Z),
        '-8.9 µm \n (autofocus) ': ('k', man_focus_0Z),
    '-11.8 µm': ('y', man_focus_n28Z),
    '-14.1 µm': ('grey', man_focus_n52Z)
}


fig, (ax1) = plt.subplots(1, 1, figsize = (3.7,6.5))

# Plotting loop
for label, (color, arr) in datasets.items():
    idx = np.argmin(np.abs(arr[:, 0] - 1385))
    y_offset = arr[idx, 1]
    x = arr[:, 0]
    y = arr[:, 1] - y_offset
    ax1.plot(x, y, '.-', color=color, label=label)

#ax1.legend(fontsize=8)
ax1.set_xlim([1385, 1392])
ax1.set_ylabel('Intensity (relative to bck @ 1283 cm⁻¹)')
ax1.set_xlabel('Raman shift (cm$^{-1}$)')
fig.tight_layout()
ax1.legend(fontsize=8)
plt.ylim([-50, 650])
fig.savefig('Sectioning_diad2.png', dpi=300, transparent=True)
```

## using the same color schemes

```python
import numpy as np
import matplotlib.pyplot as plt

# Define color map based on rounded offset values
offset_color_map = {
    '-14.1': 'grey',
    '-11.8': 'y',
    '+2.3': 'b',
    '0': 'magenta',
    '-3.3': 'c',
    '-6.6': 'r',
    '-8.9': 'k'
}


def get_color_from_offset(val):
    rounded = np.round(val, 1)
    if rounded == 0:
        return offset_color_map.get('0', 'k')
    key = f"{'+' if rounded > 0 else ''}{rounded}"
    return offset_color_map.get(key, 'k')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 5))

# --- ax1: Diad2 Area ---
ax1.plot(df_G7_grind['Depth'], df_G7_grind['Diad2_Voigt_Area'], '-c',lw=2, alpha=0.3)
ax1.errorbar(df_G7_grind['Depth'], df_G7_grind['Diad2_Voigt_Area'],
             fmt='+', ecolor='k', mec='k', ms=7, capsize=3)

ax1.set_xlabel('Depth below surface (µm)')
ax1.set_ylabel('Diad 2 area')

# Top x-axis: colored crosses by offset
ax1t = ax1.twiny()
for i, row in df_G7_focus.iterrows():
    color = get_color_from_offset(row['offset_from_optimal'])
    ax1t.errorbar(row['offset_from_optimal'], row['Diad2_Voigt_Area'],
                  fmt='o', ecolor='k', mfc=color, mec='k', ms=7, capsize=3)
ax1t.plot(df_G7_focus['offset_from_optimal'], df_G7_focus['Diad2_Voigt_Area'],
          '-k', lw=2, alpha=0.3)
ax1t.set_xlim([-20, 10])
ax1t.set_xlabel('Offset from optimal focus (µm)')

# --- ax2: Density Error ---
ax2t = ax2.twiny()
ax2.plot(df_G7_grind['Depth'], df_G7_grind['σ Density g/cm3'], '-c', lw=2, alpha=0.3)
ax2t.plot(df_G7_focus['offset_from_optimal'], df_G7_focus['σ Density g/cm3'], '-k', lw=2, alpha=0.3)

ax2.plot(df_G7_grind['Depth'], df_G7_grind['σ Density g/cm3'], '+k')

for i, row in df_G7_focus.iterrows():
    color = get_color_from_offset(row['offset_from_optimal'])
    ax2t.plot(row['offset_from_optimal'], row['σ Density g/cm3'],
              marker='o', mec='k', color=color, markersize=7)

ax2.set_ylabel('Error on density (g/cm$^{3}$)')
ax2.set_xlabel('Depth below surface (µm)')
ax2t.set_xlabel('Offset from optimal focus (µm) ')
ax1t.set_xlim([-10, 20])
ax2t.set_xlim([-10, 20])
plt.tight_layout()
fig.savefig('Offsets_G7_focus.png', dpi=300)
```
